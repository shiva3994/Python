### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Histogram of `age` with 20 bins, a title and axis labels.

```python
plt.hist(df["age"], bins=20)
plt.title("Age distribution")
plt.xlabel("Age")
plt.ylabel("Users")
plt.show()
```

### Q2. Bar chart: number of users per country.

```python
df["country"].value_counts().plot(kind="bar")
plt.title("Users per country")
plt.ylabel("Users")
plt.show()
```

### Q3. Horizontal bar chart of subscription type shares (%).

```python
(df["subscription_type"].value_counts(normalize=True) * 100).plot(kind="barh")
plt.title("Subscription type share (%)")
plt.show()
```

### Q4. Boxplot of listening hours by subscription type.

```python
df.boxplot(column="avg_listening_hours_per_week", by="subscription_type", rot=45)
plt.title("Listening hours by plan")
plt.suptitle("")
plt.show()
```

### Q5. Scatter plot: listening hours vs playlists created (use a random sample of 2000 rows and `alpha=0.3`).

```python
s = df.sample(2000, random_state=1)
plt.scatter(s["avg_listening_hours_per_week"], s["playlists_created"], alpha=0.3)
plt.xlabel("Hours per week")
plt.ylabel("Playlists")
plt.title("Hours vs playlists")
plt.show()
```

### Q6. Line chart: signups per year.

```python
yr = pd.to_datetime(df["signup_date"], format="%d-%m-%Y").dt.year
yr.value_counts().sort_index().plot(kind="line", marker="o")
plt.title("Signups per year")
plt.ylabel("Users")
plt.show()
```

### Q7. Pie chart of primary device distribution.

```python
df["primary_device"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.title("Primary device")
plt.show()
```

### Q8. Bar chart of average daily skips per favorite genre, sorted, with rotated labels.

```python
df.groupby("favorite_genre")["avg_skips_per_day"].mean().sort_values().plot(kind="bar")
plt.xticks(rotation=45)
plt.ylabel("Avg skips per day")
plt.title("Skips by genre")
plt.show()
```

### Q9. Two charts side by side: histogram of listening hours and histogram of daily skips.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df["avg_listening_hours_per_week"], bins=30)
axes[0].set_title("Listening hours")
axes[1].hist(df["avg_skips_per_day"], bins=20)
axes[1].set_title("Daily skips")
plt.show()
```

### Q10. Seaborn heatmap of the correlation matrix of numeric columns, with numbers shown.

```python
import seaborn as sns
corr = df[["age", "avg_listening_hours_per_week", "playlists_created", "avg_skips_per_day", "music_suggestion_rating_1_to_5"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation heatmap")
plt.show()
```
