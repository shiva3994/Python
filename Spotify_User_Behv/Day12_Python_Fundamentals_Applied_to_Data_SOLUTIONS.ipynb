### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Create a list of unique countries and print each on its own line using a `for` loop.

```python
countries = df["country"].unique().tolist()
for c in countries:
    print(c)
```

### Q2. Turn the country counts into a dictionary and print the count for 'India'.

```python
counts = df["country"].value_counts().to_dict()
print(counts["India"])
```

### Q3. With a list comprehension, list all column names that contain the word 'inactive'.

```python
[c for c in df.columns if "inactive" in c]
```

### Q4. Write `age_category(age)` returning 'Teen' (<20), 'Adult' (20 to 49) or 'Senior' (50+). Apply it to a new column.

```python
def age_category(age):
    if age < 20:
        return "Teen"
    elif age < 50:
        return "Adult"
    return "Senior"

df["age_category"] = df["age"].apply(age_category)
df["age_category"].value_counts()
```

### Q5. Write `summarize(df, col)` that returns a dictionary with mean, median, std, min and max (rounded to 2). Call it for 3 numeric columns in a loop.

```python
def summarize(data, col):
    return {"mean": round(data[col].mean(), 2), "median": round(data[col].median(), 2), "std": round(data[col].std(), 2), "min": data[col].min(), "max": data[col].max()}

for col in ["age", "avg_listening_hours_per_week", "avg_skips_per_day"]:
    print(col, summarize(df, col))
```

### Q6. Loop over each subscription type and print: 'Premium Duo: 10.1 hours' using an f-string.

```python
for plan in df["subscription_type"].unique():
    avg = df[df["subscription_type"] == plan]["avg_listening_hours_per_week"].mean()
    print(f"{plan}: {avg:.1f} hours")
```

### Q7. Use `zip` to loop over the first 10 ages and listening hours. Print 'Heavy' if hours > 12, otherwise 'Normal'.

```python
for age, hrs in zip(df["age"].head(10), df["avg_listening_hours_per_week"].head(10)):
    label = "Heavy" if hrs > 12 else "Normal"
    print(age, hrs, label)
```

### Q8. Dictionary comprehension: genre -> average daily skips (rounded to 2).

```python
avg_skips = {g: round(df[df["favorite_genre"] == g]["avg_skips_per_day"].mean(), 2) for g in df["favorite_genre"].unique()}
avg_skips
```

### Q9. Write `filter_users(df, country, plan)` that returns matching users. Test it with India and Student.

```python
def filter_users(data, country, plan):
    return data[(data["country"] == country) & (data["subscription_type"] == plan)]

filter_users(df, "India", "Student").head()
```

### Q10. Print the top 5 countries by number of users as a numbered list (1. India - 4265 users) using `enumerate`.

```python
top5 = df["country"].value_counts().head(5)
for i, (country, n) in enumerate(top5.items(), start=1):
    print(f"{i}. {country} - {n} users")
```
