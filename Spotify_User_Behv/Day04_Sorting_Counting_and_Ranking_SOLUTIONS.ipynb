# Day 04 - Sorting, Counting and Ranking  (SOLUTIONS)

**Goal:** Answer 'which is the most / least / top N' questions.

**Quick recap (read for 2 minutes):**

- `df.sort_values('col', ascending=False)`; several columns: `sort_values(['a','b'], ascending=[True, False])`
- `df['col'].value_counts()` counts categories; `normalize=True` gives fractions
- `unique()` lists distinct values, `nunique()` counts them
- `nlargest(n, 'col')` / `nsmallest(n, 'col')` give top / bottom n rows
- `.rank(ascending=False)` assigns ranks

*There are many valid ways to solve each question. If your answer gives the same result, it is correct.*

### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Sort users by age (youngest first) and show the top 5 rows.

```python
df.sort_values("age").head()
```

### Q2. Show `user_id` and listening hours for the 10 heaviest listeners.

```python
df.sort_values("avg_listening_hours_per_week", ascending=False)[["user_id", "avg_listening_hours_per_week"]].head(10)
```

### Q3. Sort by `country` A-Z and, within each country, by `age` oldest first. Show 10 rows.

```python
df.sort_values(["country", "age"], ascending=[True, False]).head(10)
```

### Q4. Count users per country.

```python
df["country"].value_counts()
```

### Q5. Show the share (%) of each subscription type, rounded to 1 decimal.

```python
(df["subscription_type"].value_counts(normalize=True) * 100).round(1)
```

### Q6. List the unique favorite genres and how many there are.

```python
print(df["favorite_genre"].unique())
print(df["favorite_genre"].nunique())
```

### Q7. Which 5 features are most commonly the `most_liked_feature`?

```python
df["most_liked_feature"].value_counts().head(5)
```

### Q8. Use `nlargest` to get the 5 users with the most playlists.

```python
df.nlargest(5, "playlists_created")
```

### Q9. Find the 5 users with the LOWEST listening hours, ignoring users with 0 hours.

```python
df[df["avg_listening_hours_per_week"] > 0].nsmallest(5, "avg_listening_hours_per_week")
```

### Q10. Create a column `skip_rank` where the user with the most skips gets rank 1.

```python
df["skip_rank"] = df["avg_skips_per_day"].rank(ascending=False, method="min")
df[["user_id", "avg_skips_per_day", "skip_rank"]].sort_values("skip_rank").head()
```
