**Goal:** Convert text to dates and extract year, month, tenure.

**Quick recap (read for 2 minutes):**

- `pd.to_datetime(col, format='%d-%m-%Y')` (day-month-year)
- `.dt.year`, `.dt.month`, `.dt.day_name()`, `.dt.month_name()`
- Subtract dates: `(ref_date - col).dt.days`
- `pd.Timestamp('2026-03-31')` creates a fixed date
- `col.dt.to_period('M')` groups by month

### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Convert `signup_date` (format day-month-year) to a real datetime.

```python
df["signup_date"] = pd.to_datetime(df["signup_date"], format="%d-%m-%Y")
df["signup_date"].dtype
```

### Q2. Add `signup_year`.

```python
df["signup_date"] = pd.to_datetime(df["signup_date"], format="%d-%m-%Y")
df["signup_year"] = df["signup_date"].dt.year
df["signup_year"].head()
```

### Q3. Add `signup_month` with the month name (January, February, ...).

```python
df["signup_month"] = df["signup_date"].dt.month_name()
df["signup_month"].head()
```

### Q4. Find the earliest and latest signup dates.

```python
df["signup_date"].agg(["min", "max"])
```

### Q5. How many users signed up each year? Sort by year.

```python
df["signup_year"].value_counts().sort_index()
```

### Q6. Create `tenure_days` = days between signup and 2026-03-31.

```python
ref = pd.Timestamp("2026-03-31")
df["tenure_days"] = (ref - df["signup_date"]).dt.days
df["tenure_days"].describe()
```

### Q7. Create `tenure_years` rounded to 1 decimal.

```python
df["tenure_days"] = (pd.Timestamp("2026-03-31") - df["signup_date"]).dt.days
df["tenure_years"] = (df["tenure_days"] / 365.25).round(1)
df["tenure_years"].head()
```

### Q8. Which weekday has the most signups?

```python
df["signup_date"].dt.day_name().value_counts()
```

### Q9. Show users who signed up in 2023 and have a Premium Individual plan.

```python
df[(df["signup_date"].dt.year == 2023) & (df["subscription_type"] == "Premium Individual")]
```

### Q10. Show the number of signups per month for the last 12 months in the data.

```python
df.groupby(df["signup_date"].dt.to_period("M")).size().tail(12)
```
