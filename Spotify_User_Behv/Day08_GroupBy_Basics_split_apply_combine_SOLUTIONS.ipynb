# Day 08 - GroupBy Basics (split - apply - combine)  (SOLUTIONS)

**Goal:** Answer 'what is the average X for each Y' questions. The single most important analysis tool.

**Quick recap (read for 2 minutes):**

- `df.groupby('col')['value'].mean()`
- Other aggregations: `sum()`, `count()`, `size()`, `median()`, `max()`, `min()`
- Select several value columns: `df.groupby('col')[['a','b']].mean()`
- `.sort_values(ascending=False)` ranks the result
- `.idxmax()` returns the group label with the largest value

*There are many valid ways to solve each question. If your answer gives the same result, it is correct.*

### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Average weekly listening hours per subscription type.

```python
df.groupby("subscription_type")["avg_listening_hours_per_week"].mean()
```

### Q2. Number of users per primary device using `groupby` and `size()`.

```python
df.groupby("primary_device").size()
```

### Q3. Average daily skips per favorite genre, highest first.

```python
df.groupby("favorite_genre")["avg_skips_per_day"].mean().sort_values(ascending=False)
```

### Q4. Total playlists created per country.

```python
df.groupby("country")["playlists_created"].sum()
```

### Q5. Average age for Active vs Inactive users.

```python
df.groupby("subscription_status")["age"].mean()
```

### Q6. Median listening hours per device.

```python
df.groupby("primary_device")["avg_listening_hours_per_week"].median()
```

### Q7. Maximum daily skips in each country.

```python
df.groupby("country")["avg_skips_per_day"].max()
```

### Q8. Average suggestion rating per subscription type, rounded to 2 decimals.

```python
df.groupby("subscription_type")["music_suggestion_rating_1_to_5"].mean().round(2)
```

### Q9. Average hours, skips and playlists per subscription type in ONE table.

```python
df.groupby("subscription_type")[["avg_listening_hours_per_week", "avg_skips_per_day", "playlists_created"]].mean().round(2)
```

### Q10. Which country has the highest average listening hours?

```python
df.groupby("country")["avg_listening_hours_per_week"].mean().idxmax()
```
