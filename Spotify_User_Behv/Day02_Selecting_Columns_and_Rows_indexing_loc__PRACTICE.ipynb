# Day 02 - Selecting Columns and Rows (indexing, loc, iloc)  (PRACTICE)

**Goal:** Learn to pull exactly the columns and rows you want.

**Quick recap (read for 2 minutes):**

- `df['col']` -> one column (Series); `df[['a','b']]` -> several columns (DataFrame)
- `df.iloc[row_pos, col_pos]` selects by POSITION (end is excluded)
- `df.loc[row_label, col_name]` selects by LABEL (end is included)
- `df.select_dtypes(include='number')` picks columns by type
- `.copy()` makes an independent copy so changes don't touch the original
