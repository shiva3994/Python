# Day 03 - Filtering Rows with Conditions  (PRACTICE)

**Goal:** Boolean masks are the most-used skill in data analysis. Repeat until automatic.

**Quick recap (read for 2 minutes):**

- `df[df['col'] == 'x']` keeps rows where the condition is True
- Combine with `&` (and), `|` (or), `~` (not) -> wrap each condition in ( )
- `.isin([...])` for membership, `.between(a, b)` for ranges
- `df.query("col > 5 and other == 'x'")` is a readable alternative
- `(condition).sum()` counts matches, `(condition).mean()` gives the fraction
