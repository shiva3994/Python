**Goal:** Summarise numbers the way a data analyst does.

**Quick recap (read for 2 minutes):**

- `mean()`, `median()`, `std()`, `min()`, `max()`, `sum()`, `count()`
- `quantile(0.9)` or `quantile([0.25, 0.5, 0.75])` for percentiles
- `mode()` returns the most frequent value(s)
- `.agg(['min','max'])` runs several stats at once
- `df[cols].corr()` correlation matrix (-1 to +1); `a.corr(b)` for one pair

*There are many valid ways to solve each question. If your answer gives the same result, it is correct.*

### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Average age of users.

```python
df["age"].mean()
```

### Q2. Median weekly listening hours.

```python
df["avg_listening_hours_per_week"].median()
```

### Q3. Standard deviation of daily skips.

```python
df["avg_skips_per_day"].std()
```

### Q4. Get the min and max of `playlists_created` in one command.

```python
df["playlists_created"].agg(["min", "max"])
```

### Q5. Get the 25th, 50th, 90th and 95th percentiles of listening hours.

```python
df["avg_listening_hours_per_week"].quantile([0.25, 0.5, 0.9, 0.95])
```

### Q6. What is the most common favorite genre (the mode)?

```python
df["favorite_genre"].mode()[0]
```

### Q7. Run `describe()` for only age, listening hours and skips.

```python
df[["age", "avg_listening_hours_per_week", "avg_skips_per_day"]].describe()
```

### Q8. Show the correlation matrix of age, listening hours, playlists, skips and rating (rounded to 2 decimals).

```python
df[["age", "avg_listening_hours_per_week", "playlists_created", "avg_skips_per_day", "music_suggestion_rating_1_to_5"]].corr().round(2)
```

### Q9. What is the correlation between listening hours and playlists created? Is it strong?

```python
df["avg_listening_hours_per_week"].corr(df["playlists_created"])
```

### Q10. What is the average suggestion rating, and what fraction of users gave a 5?

```python
print(df["music_suggestion_rating_1_to_5"].mean())
print((df["music_suggestion_rating_1_to_5"] == 5).mean())
```
