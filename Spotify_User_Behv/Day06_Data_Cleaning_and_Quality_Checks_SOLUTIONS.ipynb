**Goal:** Real data is messy. Learn the standard checks before any analysis.

**Quick recap (read for 2 minutes):**

- `df.duplicated().sum()`, `df['id'].is_unique` check duplicates
- `astype(int)` changes a column's type
- `df.rename(columns={'old': 'new'})` renames columns
- `df.drop(columns=['col'])` removes columns
- `fillna(value)`, `dropna(subset=['col'])` handle missing values
- `df['col'].map({'Yes': 1, 'No': 0})` recodes values

### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. How many fully duplicated rows are there?

```python
df.duplicated().sum()
```

### Q2. Check whether `user_id` is unique.

```python
df["user_id"].is_unique
```

### Q3. `user_id` is stored as a decimal number. Convert it to integer.

```python
df["user_id"] = df["user_id"].astype(int)
df["user_id"].dtype
```

### Q4. Rename `music_suggestion_rating_1_to_5` to `rating`.

```python
df = df.rename(columns={"music_suggestion_rating_1_to_5": "rating"})
df.columns.tolist()
```

### Q5. Standardise all column names: lowercase, stripped of spaces.

```python
df.columns = df.columns.str.lower().str.strip()
df.columns.tolist()
```

### Q6. Create `ad_interaction_flag` = 1 for Yes and 0 for No.

```python
df["ad_interaction_flag"] = df["ad_interaction"].map({"Yes": 1, "No": 0})
df["ad_interaction_flag"].value_counts()
```

### Q7. Create `df_no_flag` which is `df` without `inactive_3_months_flag` (the original must keep the column).

```python
df_no_flag = df.drop(columns=["inactive_3_months_flag"])
print(df.shape, df_no_flag.shape)
```

### Q8. Practice with missing data: make a copy, set `age` to NaN for the first 100 rows, count the NaNs, then fill them with the median age.

```python
tmp = df.copy()
tmp.loc[:99, "age"] = np.nan
print(tmp["age"].isna().sum())
tmp["age"] = tmp["age"].fillna(tmp["age"].median())
print(tmp["age"].isna().sum())
```

### Q9. On the same kind of copy, drop the rows with missing `age` instead and print the new shape.

```python
tmp2 = df.copy()
tmp2.loc[:99, "age"] = np.nan
tmp2 = tmp2.dropna(subset=["age"])
tmp2.shape
```
