# Day 03 - Filtering Rows with Conditions  (SOLUTIONS)

**Goal:** Boolean masks are the most-used skill in data analysis. Repeat until automatic.

**Quick recap (read for 2 minutes):**

- `df[df['col'] == 'x']` keeps rows where the condition is True
- Combine with `&` (and), `|` (or), `~` (not) -> wrap each condition in ( )
- `.isin([...])` for membership, `.between(a, b)` for ranges
- `df.query("col > 5 and other == 'x'")` is a readable alternative
- `(condition).sum()` counts matches, `(condition).mean()` gives the fraction

*There are many valid ways to solve each question. If your answer gives the same result, it is correct.*

### Setup (run this first)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_user_behv.csv")
```

### Q1. Show all users from India.

```python
df[df["country"] == "India"]
```

### Q2. How many users are older than 50?

```python
(df["age"] > 50).sum()
```

### Q3. Show Premium Family users in Germany.

```python
df[(df["subscription_type"] == "Premium Family") & (df["country"] == "Germany")]
```

### Q4. Show users from USA or UK (use `|`), then do it again using `isin`.

```python
a = df[(df["country"] == "USA") | (df["country"] == "UK")]
b = df[df["country"].isin(["USA", "UK"])]
print(len(a), len(b))
```

### Q5. Show users whose subscription is NOT Free.

```python
df[df["subscription_type"] != "Free"]
```

### Q6. Show users aged 25 to 35 (inclusive) using `between`.

```python
df[df["age"].between(25, 35)]
```

### Q7. Show users whose favorite genre is Jazz, Classical or Rock.

```python
df[df["favorite_genre"].isin(["Jazz", "Classical", "Rock"])]
```

### Q8. Show Inactive users who listen less than 5 hours per week.

```python
df[(df["subscription_status"] == "Inactive") & (df["avg_listening_hours_per_week"] < 5)]
```

### Q9. Using `query()`, find Mobile users who skip more than 15 songs a day.

```python
df.query("primary_device == 'Mobile' and avg_skips_per_day > 15")
```

### Q10. What percentage of users interacted with ads?

```python
(df["ad_interaction"] == "Yes").mean() * 100
```
