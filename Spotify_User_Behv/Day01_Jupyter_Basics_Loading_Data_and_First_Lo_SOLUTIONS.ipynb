# Day 01 - Jupyter Basics, Loading Data and First Look  (SOLUTIONS)

**Goal:** Get comfortable with Jupyter and learn the 10 commands you will run on EVERY new dataset.

**Quick recap (read for 2 minutes):**

- `pd.read_csv('file.csv')` loads a CSV into a DataFrame (a table)
- `df.head()` / `df.tail()` / `df.sample()` peek at rows
- `df.shape`, `df.columns`, `df.dtypes`, `df.info()` describe structure
- `df.describe()` gives summary statistics, `df.isna().sum()` counts missing values
- Jupyter: `Shift+Enter` runs a cell, `B` adds a cell below, `M` = markdown, `Y` = code
